```python
def soma(x,y):
    resultado = x + y
    return resultado
```

```python
def diferenca(x,y):
    resultado = x - y
    return resultado
```

```python
def multi(x,y):
    resultado = x * y
    return resultado
```

```python
def div(x,y):
    if y == 0:
        resultado = "Você não pode dividir por zero"
    else:
        resultado = x / y
    return resultado
```
