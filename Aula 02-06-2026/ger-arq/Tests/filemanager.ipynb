# Processamento de arquivos

```python
# Leitura de arquivos

arquivo001 = open("arquivo001.txt", 'r')



arquivo001.close()
```

```python
arquivo002 = open("arquivo002.txt", "w")
for linha in range(1, 101):
#    arquivo002.write('%d\n' % (linha+1))
    arquivo002.write(f'{linha}\n')



arquivo002.close()
```

```python
arquivo003 = open("arquivo003.txt", 'w')

arquivo003.write('casa')

arquivo003.close()
```

## Um outro jeito de usar a função open - gerenciador de contexto with

```python
with open("arquivo004.txt", 'w') as arquivo004obj:
    arquivo004obj.write('casa')
```

## Criando um csv

```python
with open('arquivo005.csv','w') as arquivo005:
    i = 0
    while i < 11 :
        arquivo005.write(f'{i}, {2*i}\n')
        i += 1
```
