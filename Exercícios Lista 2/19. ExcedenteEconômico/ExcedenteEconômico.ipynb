Crie uma classe que calcule numericamente o Excedente do Consumidor e o Excedente do Produtor a partir de curvas lineares de
oferta e demanda em um ponto de equilíbrio.

```python
class ExcedenteEconomico:
    def __init__ (self, demanda_lin: float, demanda_ang: float, oferta_lin: float, oferta_ang: float) -> None:
        self.d_lin = demanda_lin
        self.d_ang = demanda_ang
        self.o_lin = oferta_lin
        self.o_ang = oferta_ang
    def consumidor(self) -> float:
        p_zero = self.d_lin
        q_eq = (self.d_lin - self.o_lin) / (self.o_ang + self.d_ang)
        p_eq = self.d_lin - self.d_ang*q_eq
        excedente = ((p_zero - p_eq)*q_eq)/2
        return excedente
    def produtor(self) -> float:
        p_zero = self.o_lin
        q_eq = (self.d_lin - self.o_lin) / (self.o_ang + self.d_ang)
        p_eq = self.d_lin - self.d_ang*q_eq
        excedente = ((p_eq - p_zero)*q_eq)/2
        return excedente
```

```python
teste = ExcedenteEconomico(5, 1, 1, 0.5)
```
