Crie uma classe que simule um cronômetro com métodos para iniciar, pausar e resetar o tempo, armazenando internamente o estado atual.

```python
import time

class Cronometro:
    def __init__ (self) -> None:
        self.current_time = 0.00
        self.init_time = 0.00
        self.final_time = 0.00
    def iniciar (self) -> None:
        self.init_time = time.perf_counter()
        print("Cronômetro iniciado!")
    def pausar (self) -> None:
        self.final_time = time.perf_counter()
        self.current_time += self.final_time - self.init_time
        print(f"Cronômetro parado! Tempo atual: {self.current_time}")
    def resetar (self) -> None:
        self.current_time = 0.00
```

```python
teste = Cronometro()
```

```python
teste.iniciar()
```

```python
teste.pausar()
```

```python
teste.pausar()
```

```python
teste.resetar()
```
