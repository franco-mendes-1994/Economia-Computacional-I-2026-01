Receba a propensão marginal a consumir e aalíquota tributária da economia. Calcule o multiplicador e preveja o impacto final
no PIB dado um aumento autônomo em G.

```python
class MultiplicadorKeynesiano:
    def __init__ (self, c1: float, t: float) -> None:
        self.c1 = c1
        self.t = t
    def multiplicador(self) -> float:
        multiplicador = 1/(1 - self.c1*(1 - self.t))
        return multiplicdor
    def choque_g(self, choque: float) -> float:
        impacto = choque*self.multiplicador()
        return impacto
```
