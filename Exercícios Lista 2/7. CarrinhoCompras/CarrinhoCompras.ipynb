Crie uma classe que receba uma lista de itens e seus
respectivos preços. Adicione um método para calcular o valor total da compra.

```python
class CarrinhoCompras:
    def __init__ (self, produtos: list, precos: list) -> None:
        self.produtos = produtos
        self.precos = precos
    def valor(self) -> float:
        total = 0
        for i in range(len(self.produtos)):
            total += self.produtos[i] * self.precos[i]
        return total
```

```python
teste = CarrinhoCompras([10,4,18], [2,3,4])
```

```python
teste.valor()
```
