Crie uma classe que guarde a temperatura corrente em graus Celsius. Adicione métodos para converter e retornar esse valor em Fahrenheit e Kelvin.

```python
class Termometro:
    def __init__ (self, temperatura: float) -> None:
        self.temp = temperatura
    def farenheit(self) -> float:
        result = (self.temp*(9/5)) + 32
        return result
    def kelvin(self) -> float:
        result = self.temp + 273.15
        return result
```

```python
teste = Termometro(0)

teste.farenheit()
```
