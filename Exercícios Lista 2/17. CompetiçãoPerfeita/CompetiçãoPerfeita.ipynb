Uma firma opera em competição perfeita sob um preço de mercado P e possui Custo Marginal CMg = 2q. Crie um método para definir a quantidade ideal que maximiza o lucro da firma.

```python
class CompeticaoPerfeita:
    def __init__ (self, preco: float) -> None:
        self.preco = preco
    def q_max(self) -> float:
        q_max = self.preco / 2
        return q_max
        
```

```python
teste = CompeticaoPerfeita(10)
```
