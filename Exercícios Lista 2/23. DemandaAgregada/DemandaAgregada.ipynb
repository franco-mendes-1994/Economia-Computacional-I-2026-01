Integre as modelagens anteriores de consumo, investimento e gastos do governo (G) em uma única classe. Crie um método para
determinar o PIB de equilíbrio em uma economia fechada.

```python
class DemandaAgregada:
    def __init__ (self, c0: float, c1: float, i0: float, b: float, g: float, t: float) -> None:
        self.c0 = c0
        self.c1 = c1
        self.i0 = i0
        self.b = b
        self.g = g
        self.t = t
    def pib (self, i: float) -> float:
        aux_1 = 1/(1 - self.c1)
        aux_2 = self.c0 + self.i0 + self.b*i + self.g - self.c1*self.t
        pib = aux_1*aux_2
        return pib
```
