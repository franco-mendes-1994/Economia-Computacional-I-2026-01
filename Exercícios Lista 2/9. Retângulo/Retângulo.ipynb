Crie uma classe com os atributos base e altura. Implemente métodos específicos para calcular a área e o perímetro do objeto.

```python
class Retangulo:
    def __init__ (self, base: float, altura: float) -> None:
        self.base = base
        self.altura = altura
    def area(self) -> float:
        a = self.base * self.altura
        return a
    def perimetro(self) -> float:
        a = 2*self.base + 2*self.altura
        return a
```

```python
teste = Retangulo(3,3)
```

```python
teste.area()
```

```python
teste.perimetro()
```
