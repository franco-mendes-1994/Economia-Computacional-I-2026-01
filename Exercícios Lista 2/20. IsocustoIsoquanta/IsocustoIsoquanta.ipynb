Modele as restrições de custo de produção da firma dada a contratação de capital (K) e trabalho (L). Implemente o cálculo da Taxa Marginal de Substituição Técnica (TMST).

```python
class IsocustoIsoquanta:
    def __init__ (self, r: float, w: float, alfa: float, beta: float) -> None:
        self.r = r
        self.w = w
        self.alfa = alfa
        self.beta = beta
    def tmst(self, k: float, l: float) -> float:
        tmst = (self.alfa/self.beta)*(l/k)
        return tmst
```
