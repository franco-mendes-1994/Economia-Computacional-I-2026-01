Modele a demanda por moeda por motivos de transação e especulação: L(Y, i) = kY − hi. Calcule a demanda por liquidez informando o nível do PIB e a taxa de juros corrente.

```python
class DemandaMoeda:
    def __init__ (self, k: float, h: float) -> None:
        self.k = k
        self.h = h
    def demanda_moeda (self, y: float, i: float) -> float:
        demanda_moeda = y*self.k - i*self.h
        return demanda_moeda
```

```python
teste = DemandaMoeda(0.5, 0.6)
```

```python
teste.demanda_moeda(1000, 0.15)
```
