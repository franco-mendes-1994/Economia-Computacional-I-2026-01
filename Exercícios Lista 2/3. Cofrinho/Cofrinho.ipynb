Crie uma classe que armazene moedas de Real. Adicione métodos para depositar moedas de valores específicos e verificar o saldo total acumulado.

```python
class Cofrinho:
    def __init__ (self, valor_guardado: float) -> None:
        self.valor_guardado = valor_guardado
    def depositar(self, valor) -> None:
        self.valor_guardado += valor
        print(self.valor_guardado)
        
```

```python
teste = Cofrinho(0)

teste.depositar(100)
```
