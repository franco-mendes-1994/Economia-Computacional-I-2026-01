Modele a curva de demanda Qd = a−bP . Crie métodos para retornar a quantidade demandada dado um preço P, e o preço máximo que o
consumidor aceita pagar (preço de reserva).

```python
class DemandaLinear:
    def __init__ (self, a: float, b: float) -> None:
        self.a = a
        self.b = b
    def demanda(self, preco: float) -> float:
        q_demandada = self.a - (self.b*preco)
        return q_demandada
    def max_preco(self) -> float:
        max_preco = self.a / self.b
        return max_preco
```

```python
teste = DemandaLinear(5, 2)
```

```python
teste.demanda(2.5)
```

```python
teste.max_preco()
```
