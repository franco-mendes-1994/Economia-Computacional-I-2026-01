Crie uma classe com nome, número do cartão de identificação e uma lista de notas. Adicione um método para calcular e retornar a média final.

```python
class EstudanteUFRGS:
    def __init__ (self, nome: str, cartao: str, notas: list) -> None:
        self.nome = nome
        self.cartao = cartao
        self.notas = notas
    def media(self) -> float:
        return sum(self.notas) / len(self.notas)
        
```
