Crie uma classe para funcionários que registre o horário de entrada e de saída, calculando o total de horas trabalhadas no dia.

```python
class RegistroPonto:
    def __init__ (self, funcionario_id: str) -> None:
        self.funcionario_id = funcionario_id
        self.entrada = 0
        self.saida = 0
    def registrar_entrada(self) -> None:
        self.entrada = float(input("Informe o horário de entrada: "))
    def registrar_saida(self) -> None:
        self.saida = float(input("Informe o horário de saida: "))
    def total_horas(self) -> float:
        a = self.saida - self.entrada
        return a
```

```python
teste = RegistroPonto("Franco Mendes")
```

```python
teste.registrar_entrada()
```

```python
teste.registrar_saida()
```

```python
teste.total_horas()
```
