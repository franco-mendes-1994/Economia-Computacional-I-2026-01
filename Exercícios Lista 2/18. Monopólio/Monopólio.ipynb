Um monopolista possui custos totais CT = c · q e enfrenta a curva de demanda P = a−bq. Crie métodos para encontrar a quantidade e o preço que maximizam o lucro do monopólio.

```python
class Monopolio:
    def __init__ (self, a: float, b:float, c:float) -> None:
        self.a = a
        self.b = b
        self.c = c
    def max_q(self) -> float:
        max_q = (self.a - self.c) / 2*self.b
        return max_q
    def max_p(self) -> float:
        max_p = (self.a + self.c) / 2
        return max_p
```

```python
teste = Monopolio(3,4,1)
```

```python
teste.max_q()
```

```python
teste.max_p()
```
