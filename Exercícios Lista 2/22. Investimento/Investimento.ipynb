Modele a função de investimento privado como I = I0−b· i, onde i representa a taxa de juros. Calcule o volume de investimento total para diferentes taxas de juros passadas por parâmetro.

```python
class Investimento:
    def __init__ (self, i0: float, b: float) -> None:
        self.i0 = i0
        self.b = b
    def investimento(self, r: float) -> float:
        i = self.i0 - self.b*r
        return i
```

```python
teste = Investimento(200, 0.2)
```

```python
teste.investimento(0.15)
```
