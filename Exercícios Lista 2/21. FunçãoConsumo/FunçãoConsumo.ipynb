Modele a função de consumo keynesiana clássica C = C0+cY d. O método principal deve receber a renda total e os impostos diretos para retornar o consumo agregado.

```python
class FuncaoConsumo:
    def __init__ (self, c0: float, c1: float) -> None:
        self.c0 = c0
        self.c1 = c1
    def consumo (self, renda: float, impostos: float) -> float:
        renda_disponivel = renda - impostos
        consumo = renda_disponivel*self.c1 + self.c0
        return consumo
```

```python
teste = FuncaoConsumo(100, 0.8)
```

```python
teste.consumo(1000, 0.3)
```
