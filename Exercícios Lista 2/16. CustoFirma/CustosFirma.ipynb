Modele a estrutura de custos de uma firma recebendo o Custo Fixo (CF) e uma função para o Custo Variável (CV ). Calcule as funções de CT, CFMe, CVMe e CMe para uma quantidade q.

```python
class CustosFirma:
    def __init__ (self, cf: float, cv) -> None:
        self.cf = cf
        self.cv = cv
    def custos(self, q) -> float:
        total = self.cf + self.cv(q)
        cfme = self.cf / q
        cvme = self.cv(q) / q
        cme = cfme + cvme
        return [total, cfme, cvme, cme]
```

```python
def custo_variavel(q):
    return 3*q
```

```python
teste = CustosFirma(100, custo_variavel)
```

```python
teste.custos(100)
```
