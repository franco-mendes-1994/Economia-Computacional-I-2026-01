Receba a oferta real de moeda (M/P) e os parâmetros estruturais da demanda por moeda. Crie um método que retorne a taxa de juros i de equilíbrio monetário para cada nível de renda Y .

```python
class CurvaLM:
    def __init__ (self, oferta: float, k: float, h: float) -> None:
        self.oferta = oferta
        self.k = k
        self.h = h
    def juros (self, renda: float) -> float:
        juros = (1/self.h)*(self.k*renda - self.oferta)
        return juros
```

```python
teste = CurvaLM(1000, 2, 7)
```
