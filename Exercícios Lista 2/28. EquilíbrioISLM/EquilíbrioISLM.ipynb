Crie uma classe agregadora que receba um objeto do tipo CurvaIS e um objeto do tipo CurvaLM. Implemente um método que resolva o
sistema linear e retorne o par de equilíbrio simultâneo (Y ∗, i∗).

```python
class FuncaoConsumo:
    def __init__ (self, c0: float, c1: float) -> None:
        self.c0 = c0
        self.c1 = c1
    def consumo (self, renda: float, impostos: float) -> float:
        renda_disponivel = renda - impostos
        consumo = renda_disponivel*self.c1 + self.c0
        return consumo

class Investimento:
    def __init__ (self, i0: float, b: float) -> None:
        self.i0 = i0
        self.b = b
    def investimento(self, r: float) -> float:
        i = self.i0 - self.b*r
        return i

class CurvaIS:
    def __init__ (self, consumo_obj, investimento_obj, g: float, t: float) -> None:
        self.consumo = consumo_obj
        self.investimento = investimento_obj
        self.g = g
        self.t = t
    def pib (self, i: float) -> float:
        numerador = self.consumo.c0 + self.investimento.i0 + self.investimento.b*i + self.g - self.consumo.c1*self.t
        denominador = 1 - self.consumo.c1
        pib = numerador/denominador
        return pib
```

```python
class CurvaLM:
    def __init__ (self, oferta: float, k: float, h: float) -> None:
        self.oferta = oferta
        self.k = k
        self.h = h
    def juros (self, renda: float) -> float:
        juros = (1/self.h)*(self.k*renda - self.oferta)
        return juros
```
