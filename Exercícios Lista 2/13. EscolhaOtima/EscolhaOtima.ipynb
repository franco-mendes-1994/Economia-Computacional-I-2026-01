Utilizando a lógica da utilidade Cobb-Douglas do exercício anterior, crie métodos que calculem analiticamente as demandas ótimas de x e y dada uma renda e os preços de mercado.

```python
class EscolhaOtima:
    def __init__(self, alfa: float, beta: float) -> None:
        self.alfa = alfa
        self.beta = beta
    def x_otimo(self, renda: float, preco1: float) -> float:
        aux1 = renda*self.alfa
        aux2 = (self.alfa + self.beta)*preco1
        return aux1/aux2
    def y_otimo(self, renda: float, preco2: float) -> float:
        aux1 = renda*self.beta
        aux2 = (self.alfa + self.beta)*preco2
        return aux1/aux2
```

```python
teste = EscolhaOtima(0.5,0.5)
```

```python
teste.x_otimo(1000, 10)
```

```python
teste.y_otimo(1000, 5)
```
