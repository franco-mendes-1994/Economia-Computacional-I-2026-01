Modele a função de utilidade U = xα · yβ. Crie um método para calcular a utilidade de uma cesta e outro para calcular a Taxa Marginal de Substituição (TMS).

```python
class UtilidadeCobbDouglas:
    def __init__ (self, alfa: float, beta: float) -> None:
        self.bem1 = None
        self.bem2 = None
        self.alfa = alfa
        self.beta = beta
    def utilidade(self, bem1: float, bem2: float) -> float:
        self.bem1 = bem1
        self.bem2 = bem2
        u = (self.bem1**self.alfa)*(self.bem2**self.beta)
        return u
    def tms(self) -> float:
        aux1 = self.bem2/self.bem1
        aux2 = self.alfa/self.beta
        tms = aux1*aux2
        return f"A taxa marginal de substituição do bem 1 pelo bem 2 é {tms}"
```

```python
teste = UtilidadeCobbDouglas(0.5, 0.5)
```

```python
teste.utilidade(10, 5)
```
