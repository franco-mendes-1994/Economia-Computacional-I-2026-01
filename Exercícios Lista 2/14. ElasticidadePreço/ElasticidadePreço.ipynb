Crie uma classe que receba dados de preço e quantidade demandada em dois momentos distintos. Implemente o cálculo da elasticidade preço da demanda pelo método do arco.

```python
class ElasticidadePreco:
    def __init__ (self, preco: list, quantidade: list) -> None:
        self.preco = preco
        self.quant = quantidade
    def elasticidade_preco(self) -> float:
        var_quant = self.quant[1] - self.quant[0]
        var_preco = self.preco[1] - self.preco[0]
        aux1 = var_quant / self.quant[1]
        aux2 = var_preco / self.preco[1]
        return aux1 / aux2
```
