Crie uma classe com os atributos titular, saldo e limite.
Implemente os métodos depositar e sacar (garantindo que o saque respeite o limite
disponível).

```python
class ContaBancaria:
    def __init__ (self, titular: str, saldo: float, limite: float) -> None:
        self.titular = titular
        self.saldo = saldo
        self.limite = limite
    def depositar(self, valor) -> None:
        self.saldo += valor
    def sacar(self, valor) -> None: 
        if (self.saldo - valor) < (0 - self.limite):
            print("Valor indisponível para saque. Limite esgotado")
        else:
            self.saldo -= valor
        
```

```python
teste = ContaBancaria("Franco", 500, 200)
```

```python
teste.depositar(200)
```
