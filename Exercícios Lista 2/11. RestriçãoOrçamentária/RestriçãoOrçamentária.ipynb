Receba a renda (m), o preço do bem 1 (p1) e o preço do bem 2 (p2). Crie um método que avalie se uma cesta genérica (x1, x2) é
financeiramente viável.

```python
class RestricaoOrcamentaria:
    def __init__ (self, renda: float, precos: list) -> None:
        self.renda = renda
        self.precos = precos
    def avaliar(self, x: list) -> bool:
        cost = x[0]*self.precos[0] + x[1]*self.precos[1]
        if cost > self.renda:
            return False
        else:
            return True
```

```python
teste = RestricaoOrcamentaria(1000, [10 , 5])
```

```python
teste.avaliar([100, 100])
```

```python
teste.avaliar([50,20])
```
